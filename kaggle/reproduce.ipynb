# Density transport: RSNA post hoc analyses and CBIS-DDSM prediction export
Clones github.com/aitlbachir/density-transport and runs `experiments/rsna_posthoc.py` on the 512 and 1024 px
training runs, then collects the CBIS-DDSM predictions of the five training runs. Internet on, CPU.

```python
import glob, os, subprocess, sys
print("cloning", flush=True)
subprocess.run("git clone -q https://github.com/aitlbachir/density-transport /kaggle/working/density-transport", shell=True, check=True, timeout=300)
os.chdir("/kaggle/working/density-transport")
print(subprocess.run("git log -1 --format=%H", shell=True, capture_output=True, text=True).stdout, flush=True)
runs = sorted({os.path.dirname(p) for p in glob.glob("/kaggle/input/**/oof_breast.csv", recursive=True)})
print(runs, flush=True)
for d in runs:
    label = "1024" if "fork-of" in d else "512"
    r = subprocess.run([sys.executable, "experiments/rsna_posthoc.py", d, label, "/kaggle/working"], capture_output=True, text=True)
    print(label, r.stdout[:100], r.stderr[-3000:], flush=True)
for f in sorted(glob.glob("/kaggle/working/rsna_posthoc_*.json")):
    print("BEGIN", os.path.basename(f)); print(open(f).read().replace("\n", "")); print("END", os.path.basename(f), flush=True)
r = subprocess.run([sys.executable, "kaggle/export_cbis_predictions.py", "/kaggle/input", "/kaggle/working/cbis_raw_predictions.csv"], capture_output=True, text=True)
print("BEGIN cbis_raw_predictions.csv"); print(r.stdout); print(r.stderr[-3000:]); print("END cbis_raw_predictions.csv", flush=True)
```
